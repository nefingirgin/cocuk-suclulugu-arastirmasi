```python
suclar = pd.read_csv('/content/cocuk_suclulugu_suc_turleri.csv.csv', sep=';')
print(suclar)
```

```python
for sutun in ["yil_2024", "yil_2025"]:
    suclar[sutun] = suclar[sutun].str.replace(",", ".").astype(float)

print(suclar)
```

```python
import matplotlib.pyplot as plt

suclar.plot(
    x="suc_turu",
    y=["yil_2024", "yil_2025"],
    kind="bar",
    figsize=(10, 6)
)

plt.title("2024-2025 Yıllarında Suç Türlerinin Karşılaştırılması")
plt.xlabel("Suç Türü")
plt.ylabel("Oran (%)")
plt.xticks(rotation=20)
plt.legend(["2024", "2025"])
plt.tight_layout()

plt.savefig("suc_turleri_karsilastirma.png")
plt.show()
```

```python
suclar.to_csv("cocuk_suclulugu_suc_turleri_analiz.csv", index=False)

print("Suç türleri verisi kaydedildi.")
```
